This notebook reproduces Figure 4 in "Humidity’s role in population heat-health risk" by Kong et al.

```python
"""
import python packages
please install required packages before running this notebook
"""
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy.stats import gaussian_kde
from matplotlib.lines import Line2D
%matplotlib inline
%config InlineBackend.figure_format = "retina"

# Nature Communications artwork requirements: sans-serif font, preferably Helvetica/Arial
# (Liberation Sans is metrically identical to Arial); text 5-7 pt at final print size;
# vector PDF with true (editable) embedded fonts, not the Type 3 fonts matplotlib embeds
# by default
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Liberation Sans', 'Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['mathtext.fontset'] = 'custom'
plt.rcParams['mathtext.rm'] = 'Liberation Sans'
plt.rcParams['mathtext.it'] = 'Liberation Sans:italic'
plt.rcParams['mathtext.bf'] = 'Liberation Sans:bold'
```

```python
# read in mean absolute prediction error for GHSI
GHSI_mae=xr.open_dataset('../data/GHSI_MAE.nc').MAE
# scale from daily to summer (122 days during June-Sep) total in 10^3
GHSI_mae=GHSI_mae*122/1000

# read in mean absolute prediction error for existing heat stress indices
HSI_mae=xr.open_dataset('../data/HSI_MAE.nc').MAE
# scale from daily to summer (122 days during June-Sep) total in 10^3
HSI_mae=HSI_mae*122/1000

# read in Japan country-level effective beta of existing heat stress indices
HSI_eff_beta=xr.open_dataset('../data/Country_level_HSI_effective_beta.nc').beta
```

```python
# create figure
fig = plt.figure(figsize=(7.087, 4.387))  # 180 x 111 mm: NC 2-column width, same
                                           # aspect ratio as the original 10.5x6.5 design
ax1 = fig.add_subplot(2,1,1)  
ax2 = fig.add_subplot(2,1,2)  

# panel a
for x in np.arange(1,48): # loop over prefecture index
    mae=np.log10(HSI_mae.sel(pf=x).values) # MAE for a certain prefecture; take log for better visualization
    kde = gaussian_kde(mae) # calculate kernel density of MAE
    y = np.linspace(np.min(mae), np.max(mae), 200)
    v = kde(y) # kernel density
    v = v / v.max() * 0.4 # normalize to set the width of violin plot
    ax1.fill_betweenx(y, x - v, x + v, facecolor='lightgray',edgecolor='gray', alpha=0.8, linewidth=0.5) # violin plot
    
    q25, q50, q75 = np.percentile(mae, [25, 50, 75]) # 25th, 50th, and 75th percentile of log10(MAE)
    ax1.hlines(q50, x - kde(q50)/kde(y).max()*0.4, x + kde(q50)/kde(y).max()*0.4, color='w', linewidth=1) # add horizontal line at 50th percentile
    if x==1: # add legend only for the prefecture index=1
        ax1.scatter(x,np.log10(HSI_mae.sel(pf=x).sel(HSI='WBGT')),marker='x',s=12,c='#D55E00', linewidths=1,zorder=6,label='WBGT')
        #ax1.scatter(x,np.log10(HSI_mae.sel(pf=x).sel(HSI='WBGTin')),marker='x',s=12,c='#D55E00', linewidths=0.5,zorder=6,label=r'$\mathrm{WBGT}_{\mathrm{in}}$')
        ax1.scatter(x,np.log10(HSI_mae.sel(pf=x).sel(HSI='Tw')),c='burlywood',marker='*',s=16,zorder=5,label=r'$T_{\mathrm{w}}$')
        ax1.scatter(x,np.log10(GHSI_mae.sel(pf=x).sel(exposure='indoor').min('beta')),edgecolors='black',marker='o',s=12,facecolors='none',linewidths=0.5,zorder=5,label=r'$\mathrm{GHSI}_{\mathrm{in}}$')
        ax1.scatter(x,np.log10(GHSI_mae.sel(pf=x).sel(exposure='outdoor').min('beta')),edgecolors='black',marker='o',s=12,facecolors='none',linewidths=1.2,zorder=5,label='GHSI')
    else:
        ax1.scatter(x,np.log10(HSI_mae.sel(pf=x).sel(HSI='WBGT')),marker='x',s=12,c='#D55E00', linewidths=1,zorder=6)
        #ax1.scatter(x,np.log10(HSI_mae.sel(pf=x).sel(HSI='WBGTin')),marker='x',s=12,c='#D55E00', linewidths=0.5,zorder=6)
        ax1.scatter(x,np.log10(HSI_mae.sel(pf=x).sel(HSI='Tw')),c='burlywood',marker='*',s=16,zorder=5)
        ax1.scatter(x,np.log10(GHSI_mae.sel(pf=x).sel(exposure='indoor').min('beta')),edgecolors='black',marker='o',s=12,linewidths=0.5, facecolors='none',zorder=5)
        ax1.scatter(x,np.log10(GHSI_mae.sel(pf=x).sel(exposure='outdoor').min('beta')),edgecolors='black',marker='o',s=12,linewidths=1.2, facecolors='none',zorder=5)
        
ax1.tick_params(axis='both', labelsize=7)
ax1.set_xticks(np.arange(1,48))
ax1.set_xticklabels([str(i) if i in [1, 10, 20, 30, 40,47] else '' for i in range(1, 48)])
ax1.set_xlim(0,48)
ax1.set_xlabel('Prefecture index',fontsize=7,labelpad=2)
ax1.set_ylabel('$\log_{10}(\mathrm{MAE})$',fontsize=7)

# panel 2
X=GHSI_mae.sel(exposure='indoor').sum('pf').sel(beta=slice(-4,16.5))['beta'].values # X variable
# plot country total MAE of indoor GHSI across a wide range of beta values
ax2.plot(X,GHSI_mae.sel(exposure='indoor').sum('pf').sel(beta=slice(-4,16.5)),zorder=0,c='k',linewidth=1,label=r'$\mathrm{GHSI}_{\mathrm{in}}$ MAE ($\beta$)')
# below we plot scatters representing the effective beta (X-variable) and country-total MAE of existing heat stress indices
# we found that these scatters lie close to the indoor GHSI MAE curve.
# This suggests that the predictive performance of existing heat stress indices are well preserved after linearization 
ax2.scatter(HSI_eff_beta.sel(var='HI'),HSI_mae.sum('pf').sel(HSI='HI'),c='blue') # heat index
ax2.scatter(HSI_eff_beta.sel(var='Ts'),HSI_mae.sum('pf').sel(HSI='Ts'),c='gold') # leathal temperature
ax2.scatter(HSI_eff_beta.sel(var='Tw'),HSI_mae.sum('pf').sel(HSI='Tw'),c='burlywood') # wet-bulb temperature
ax2.scatter(HSI_eff_beta.sel(var='WBGTin'),HSI_mae.sum('pf').sel(HSI='WBGTin'),c='#D55E00') # WBGT indoor
ax2.scatter(HSI_eff_beta.sel(var='WBGT'),HSI_mae.sum('pf').sel(HSI='WBGT'),c='#D55E00',marker='s') # WBGT outdoor
ax2.scatter(HSI_eff_beta.sel(var='UTCIin'),HSI_mae.sum('pf').sel(HSI='UTCIin'),c='deepskyblue') # UTCI indoor
ax2.scatter(HSI_eff_beta.sel(var='UTCI'),HSI_mae.sum('pf').sel(HSI='UTCI'),c='deepskyblue',marker='s') # UTCI outdoor
ax2.scatter(HSI_eff_beta.sel(var='Hx'),HSI_mae.sum('pf').sel(HSI='Hx'),c='dimgray') # Humidex
ax2.scatter(HSI_eff_beta.sel(var='sWBGT'),HSI_mae.sum('pf').sel(HSI='sWBGT'),c='darkgray') # simplified WBGT
ax2.scatter(0.98,GHSI_mae.sel(exposure='indoor').sum('pf').min('beta'),c='black') # indoor GHSI with country-level optimal beta of 0.98C/kPa
ax2.scatter(1.56,GHSI_mae.sel(exposure='outdoor').sum('pf').min('beta'),c='black',marker='s') # outdoor GHSI with country-level optimal beta of 1.56C/kPa
ax2.scatter(0,GHSI_mae.sel(exposure='indoor').sum('pf').sel(beta=0),c='purple') # MAE of dry-bulb temperature which is equivalent to GHSI indoor with zero weight on humidity

ax2.tick_params(axis='both', labelsize=7)
ax2.set_xlim(-1,16.5)
ax2.set_ylabel('MAE '+ '('+r'$\mathrm{10}^{\mathrm{3}}$'+')',fontsize=7)
ax2.set_xlabel(r'$\beta$ (°C/kPa)',fontsize=7)
ax2.hlines(HSI_mae.sum('pf').sel(HSI='Ta'), -1,16.5,color='grey', linewidth=1,linestyle='dashed',zorder=0)
ax2.vlines(0,20,32,color='grey', linewidth=1,linestyle='dashed',zorder=0)
ax2.set_ylim(20,32)

anno_opts = dict(xy=(0.02, 0.9), xycoords='axes fraction',va='baseline', ha='center')
ax1.annotate('a', **anno_opts,fontsize=7,fontweight='bold')  
ax2.annotate('b', **anno_opts,fontsize=7,fontweight='bold')  

ax2.text(15.8,30.5,r'$T_{\mathrm{w}}$',fontsize=7,c='black')  
ax2.text(-0.4,24,r'$T_{\mathrm{a}}$',fontsize=7,c='black')
ax2.text(0.4,21.8,'HI',fontsize=7,c='black')
ax2.text(4.5,25.4,r'$T_{\mathrm{S}}$',fontsize=7,c='black')
ax2.text(6,24.4,r'$\mathrm{WBGT}_{\mathrm{in}}$',fontsize=7,c='black')
ax2.text(6,21.7,'WBGT',fontsize=7,c='black')
ax2.text(2.6,24.3,r'$\mathrm{UTCI}_{\mathrm{in}}$',fontsize=7,c='black')
ax2.text(1.7,23.7,'UTCI',fontsize=7,c='black')
ax2.text(1.75,21.0,'GHSI',fontsize=7,c='black')
ax2.text(0.5,23.7,r'$\mathrm{GHSI}_{\mathrm{in}}$',fontsize=7,c='black')
ax2.text(5.4,25.8,r'$H_{\mathrm{X}}$',fontsize=7,c='black')
ax2.text(6.4,26.6,'sWBGT',fontsize=7,c='black')


# add legend
legend_elements = [
    Line2D([0], [0], marker='o', linestyle='None',markeredgecolor='black',markeredgewidth=1.2, markerfacecolor='none', label='GHSI', markersize=8),
    Line2D([0], [0], marker='o', linestyle='None',markeredgecolor='black',markeredgewidth=0.5, markerfacecolor='none',label=r'$\mathrm{GHSI}_{\mathrm{in}}$', markersize=8),
    Line2D([0], [0], marker='x', linestyle='None',color='#D55E00',markeredgewidth=1, label='WBGT', markersize=8),
    Line2D([0], [0], marker='*', color='burlywood', linestyle='None', label=r'$T_{\mathrm{w}}$', markersize=8),
    ]

ax1.legend(
    handles=legend_elements,
    loc=(0.59, 0.75),
    frameon=False,
    fontsize=7,
    ncol=2   # <-- two columns
)

ax2.legend(fontsize=7,frameon=False,loc=(0.1,0.8))

for ax in [ax1, ax2]:
    for spine in ax.spines.values():
        spine.set_linewidth(0.5)
    ax.tick_params(width=0.5, length=2.33)

fig.subplots_adjust(bottom=0.10, left=0.075, right=0.983, top=0.965, hspace=0.2)


out_dir = '/home/kong97/coding/optimal_metric/Japan/revision/plot/publication_data_share'
fig.savefig(f'{out_dir}/Figure4_v2.pdf', bbox_inches=None)
print('Figure size (mm):', fig.get_size_inches()*25.4)
```

**Figure 4**. Prediction accuracy comparison between GHSI and existing heat stress indices. (a) Distribution of summer cumulative mean absolute error (MAE; log$_{\mathrm{10}}$ scale) across existing heat stress indices for each prefecture (violin plots; vertical bounds show minima and maxima, widths indicate probability density; white horizontal line indicates the median). Overlaid points show MAE for GHSI, GHSI$_{\mathrm{in}}$, WBGT, and $T_{\mathrm{w}}$. (b) National total summer cumulative MAE for GHSI$_{\mathrm{in}}$ as a function of $\beta$ (solid curve). Overlaid points mark existing heat stress indices, $T_{\mathrm{a}}$, GHSI and GHSI$_{\mathrm{in}}$, plotted at their national optimal/effective $\beta$ and corresponding national MAE; indices that incorporate solar radiation (GHSI, WBGT, and UTCI) are highlighted with squares.
